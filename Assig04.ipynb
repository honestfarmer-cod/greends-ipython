```python
class Jar:
    def __init__(self, capacity=12):
        if not isinstance(capacity, int):
            raise ValueError("that's not cookie")
        if capacity < 0:
            raise ValueError('no more cookie')
        self._capacity = capacity
        self._size = 0

    def __str__(self):
        return self._size * "🍪"

    def deposit(self, n):
        if not isinstance(n, int):
            raise ValueError("that aint no cookie")
        if n < 0:
            raise ValueError("no more cookie for you")
        if self._size + n > self._capacity:
            raise ValueError('that`s too much cookie')
        self._size += n

    def withdraw(self, n):
        if not isinstance(n, int):
            raise ValueError("you take air")
        if n < 0:
            raise ValueError("you take a crumb")
        if self._size - n < 0:
            raise ValueError("noooo all cookies gone")
        self._size -= n

    @property
    def capacity(self):
        return self._capacity

    @property
    def size(self):
        return self._size


```

```python
from jar import Jar

def test_init():
    jar = Jar()
    assert jar.capacity == 12
    assert jar.size == 0


def test_str():
    jar = Jar()
    assert str(jar) == ""
    jar.deposit(1)
    assert str(jar) == "🍪"
    jar.deposit(11)
    assert str(jar) == "🍪🍪🍪🍪🍪🍪🍪🍪🍪🍪🍪🍪"


def test_deposit():
    jar = Jar()
    jar.deposit(1)
    assert jar.size == 1


def test_withdraw():
    jar = Jar(capacity=5)
    jar.deposit(3)
    assert jar.size == 3
    jar.withdraw(3)
    assert jar.size == 0
```
