```python
import pandas as pd
df = pd.read_csv(f'{path}/kiva_loans.csv')
print(df)
```

```python
#1. Select rows using string methods (e.g. .isin) and numerical conditions
df[(df['sector'].isin(['Agriculture'])) & (df['loan_amount'] > 1000)]
```

```python
#2. Sort table using one column
display (df.sort_values(by='date', ascending=False))
```

```python
#2 Sort table using multiple columns
display (df.sort_values(by=['posted_time', 'loan_amount'], ascending=False ))
```

```python
#3. Group and aggregate rows with "groupby"
display(df[df['sector'] == 'Agriculture'].groupby('sector')['loan_amount'].sum().reset_index())
```

```python
#4. Create a new variable with "apply" and "lambda"
df.groupby("loan_amount", group_keys=True).apply(lambda x: x, include_groups=False)
```

```python
#5 Scaling to large datasets
selected_columns = ['id', 'loan_amount', 'sector', 'country_code', 'currency', 'partner_id', 'date']
df_selected = df[selected_columns]
display(df_selected.head())
```

```python
#6 Calculate 'unfunded_amount' as the difference between 'loan_amount' and 'funded_amount'
df = df.eval('unfunded_amount = loan_amount - funded_amount')

# 'funded_ratio' the ratio of 'funded_amount' to 'loan_amount'
df = df.eval('funded_ratio = funded_amount / loan_amount')
display(df[['loan_amount', 'funded_amount', 'unfunded_amount', 'funded_ratio']].head())
```
